# Problem 1

A physiotherapist with a male football team is interested in studying the relationship between foot injuries and the positions at which the players play from the data collected.

```python
import pandas as pd
# creating dataset
data={
    "Position": ["Striker", "Forward", "Attacking Midfielder", "Winger"],
    "Injured": [45, 56, 24, 20],
    "Not_Injured": [32, 38, 11, 9]
}
df= pd.DataFrame(data)
# Add Total column
df["Total"] = df["Injured"] + df["Not_Injured"]

df
```

```python
total_players = df["Total"].sum()
total_injured = df["Injured"].sum()
total_not_injured = df["Not_Injured"].sum()

print("Total Players:", total_players)
print("Total Injured:", total_injured)
print("Total Not Injured:", total_not_injured)
```

**1.1 The probability that a randomly chosen player would suffer an injury**

```python
p_injury = total_injured / total_players
print("Probability that a randomly chosen player would suffer an injury:", p_injury)
```

Approximately 61.7% of players suffered injuries, indicating a relatively high injury rate in the team.

**1.2 The probability that a player is a forward or a winger**

```python
p_forward_or_winger = df[df["Position"].isin(["Forward", "Winger"])]["Total"].sum() / total_players
print("Probability that a player is a forward or a winger:", p_forward_or_winger)
```

About 52.3% of players play in forward or winger positions.

**1.3 The probability that a randomly chosen player plays in a striker position and has a foot injury**

```python
p_striker_and_injured = df[df["Position"] == "Striker"]["Injured"].values[0] / total_players
print("Probability that a randomly chosen player plays in a striker position and has a foot injury:", p_striker_and_injured)
```

Around 19.1% of all players are injured strikers.

**1.4 The probability that a randomly chosen injured player is a striker.**

```python
p_striker_given_injured = df[df["Position"] == "Striker"]["Injured"].values[0] / total_injured
print("Probability that a randomly chosen injured player is a striker):", p_striker_given_injured)
```

Among injured players, 31.0% are strikers, suggesting strikers may have a relatively higher injury involvement compared to some other positions

The analysis shows a high injury prevalence (61.7%) among players, with forwards and strikers contributing the highest number of injuries, while wingers and attacking midfielders exhibit relatively higher injury rates. Therefore, implementing position-specific training, effective workload management, and improved recovery strategies is essential to reduce injuries.

# **One Sample t-test**

**3.1 Zingaro has reason to believe that the unpolished stones may not be suitable for printing. Do you think Zingaro is justified in thinking so?**
- State the null and alternate hypotheses
- Conduct the hypothesis test and compute the p-value
- Write down conclusions from the test results

Note: Consider the level of significance as 5%."

```python
print(df.columns)
```

```python
# Extract columns
unpolished = df['Unpolished']
polished = df['Treated and Polished']
```

```python
#  ONE-SAMPLE T-TEST
# ---------------------------------------------
print("\n===== Unpolished Stones Suitability =====")

# Hypotheses
print("\nH0: Mean hardness >= 150 (Suitable)")
print("H1: Mean hardness < 150 (Not Suitable)")

# Perform one-sample t-test
t_stat, p_value = stats.ttest_1samp(unpolished, 150)
```

Hypotheses for One-Sample t-test:

•	Null Hypothesis (H0): The true mean hardness of unpolished stones is equal to 150 (μ >= 150).

•	Alternative Hypothesis (Ha): The true mean hardness of unpolished stones is less than 150 (μ < 150).

Here, we are assuming a reference mean of 150 as the minimum suitability for printing, and testing if the unpolished stones fall below this threshold.


```python
from scipy import stats
# Perform one-sample t-test
t_stat, p_value_two_tailed = stats.ttest_1samp(unpolished, 150)

# Convert to one-tailed p-value
p_value_one_tailed = p_value_two_tailed / 2

# If t-stat is positive, adjust (since we want μ < 150)
if t_stat > 0:
    p_value_one_tailed = 1 - p_value_one_tailed

print("Test statistic:", t_stat)
print("p-value:", p_value_one_tailed)

alpha = 0.05

if p_value_one_tailed < alpha:
    print(f"As the p-value ({p_value_one_tailed:.4f}) is less than the level of significance (0.05), we reject the null hypothesis.")
else:
    print(f"As the p-value ({p_value_one_tailed:.4f}) is greater than the level of significance (0.05), we fail to reject the null hypothesis.")
```

Zingaro is justified.
The unpolished stones have significantly lower hardness than 150, so they are not suitable for optimal printing.

```python
import numpy as np
from scipy.stats import norm

plt.figure()
plt.hist(unpolished, bins=10, density=True)

mean_val = unpolished.mean()
std_dev = unpolished.std()
x = np.linspace(unpolished.min(), unpolished.max(), 100)
pdf = norm.pdf(x, mean_val, std_dev)
plt.plot(x, pdf)

plt.axvline(mean_val, linestyle='dashed', label=f"Mean: {mean_val:.2f}")
plt.axvline(150, linestyle='solid', color='red', label="150 threshold")

plt.legend()
plt.title("Distribution of Unpolished Stone Hardness with Normal Curve")
plt.show()
```

Since the p-value is less than the 5% significance level, we reject the null hypothesis. Zingaro is justified in their concern.

There is strong statistical evidence that the mean hardness of unpolished stones is less than 150. Therefore, unpolished stones are not suitable for printing

```python
#  TWO-SAMPLE T-TEST
# ---------------------------------------------
print("\n===== Polished vs Unpolished Stones Suitability =====")

# Hypotheses
print("\nH0: Mean hardness is same")
print("H1: Mean hardness is different")
```

**4.2 How does the hardness of implants vary depending on methods?**

**Hypotheses**

**H₀:** Mean hardness is same across methods            
**H₁**: At least one method differs

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

for alloy_type in [1, 2]: # Changed to integer values to match 'Alloy' column
    print(f"\n                 Analyzing Alloy {alloy_type} for Methods:")
    subset = dental[dental['Alloy'] == alloy_type]

    # Groups for ANOVA
    groups = []
    for method in sorted(subset['Method'].unique()):
        groups.append(subset[subset['Method'] == method]['Response'])

    # Assumptions
    print("  Assumptions:")
    # Normality (Shapiro-Wilk)
    print("    Shapiro-Wilk Test for Normality:")
    for i, group in enumerate(groups):
        stat, p = stats.shapiro(group)
        print(f"      Method {i+1}: p-value = {p:.4f}")

    # Homogeneity of Variance (Levene)
    stat, p = stats.levene(*groups)
    print(f"    Levene's Test for Homogeneity of Variance: p-value = {p:.4f}")

    # Hypothesis Test (One-way ANOVA)
    f_stat, p_val = stats.f_oneway(*groups)
    print(f"\n  One-way ANOVA results:\n    F-statistic = {f_stat:.4f}, \n    p-value = {p_val:.4f}")

    # Post-hoc (Tukey HSD)
    if p_val < 0.05:
        print("  Significant difference found in mean hardness across methods. Performing Tukey HSD:")
        tukey = pairwise_tukeyhsd(endog=subset['Response'], groups=subset['Method'], alpha=0.05)
        print(tukey)
    else:
        print("  No significant difference found in mean hardness across methods.")
```

Since the p-value is < 0.05, we reject the null hypothesis. Hence, we have enough statistical evidence to say that the mean hardness of implants differs across methods.

**4.3 What is the interaction effect between the dentist and method on the hardness of dental implants for each type of alloy?**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Function to create interaction plot
def interaction_plot(data, alloy_value):
    # Filter for specific alloy
    subset = data[data['Alloy'] == alloy_value]

    # Compute mean hardness
    means = subset.groupby(['Method', 'Dentist'])['Response'].mean().unstack()

    # Plot
    plt.figure()
    for dentist in means.columns:
        plt.plot(means.index, means[dentist], marker='o', label=f'Dentist {dentist}')

    plt.xlabel('Method')
    plt.ylabel('Mean Hardness')
    plt.title(f'Interaction Plot for Alloy {alloy_value}')
    plt.legend(title="Dentist")
    plt.grid(True)
    plt.show()

# Plot for Alloy 1
interaction_plot(dental, 1)

# Plot for Alloy 2
interaction_plot(dental, 2)
```

**4.4 How does the hardness of implants vary depending on dentists and methods together?**

**Hypothesis**:                                     
   **Null Hypothesis (H0)**: There is no interaction between Dentist and Method on implant hardness.
   
   **Alternate Hypothesis (Ha)**: There is an interaction between Dentist and Method on implant hardness.


```python
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd


for alloy_type in [1, 2]: # Changed to integer values to match 'Alloy' column
    subset = dental[dental['Alloy'] == alloy_type].copy()
    print(f"\n--- Analysis for Alloy {alloy_type} ---")

    # Conduct Hypotheses test (Two way ANOVA)
    formula = 'Response ~ C(Dentist) + C(Method) + C(Dentist):C(Method)'
    model = ols(formula, data=subset).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)
    print(f"\n--- ANOVA Table for Alloy {alloy_type} ---")
    print(anova_table)

    #Extract P-value for the interaction term
    interaction_p_value = anova_table.loc['C(Dentist):C(Method)', 'PR(>F)']
    residuals=model.resid

    # Normality (Shapiro-Wilk)
    stat_shapiro, p_value_shapiro = stats.shapiro(residuals)
    print(f"   Normality (Shapiro-Wilk) on Residuals: p-value = {p_value_shapiro:.4f}")

    #Homogeneity of Variances (Levene) on residuals grouped by interaction term
    subset['Interaction_Group'] = subset['Dentist'].astype(str) + '_' + subset['Method'].astype(str)
    groups_for_levene= []
    for group_name in subset['Interaction_Group'].unique():
      groups_for_levene.append(subset[subset['Interaction_Group'] == group_name]['Response'])

    # Perform Levene's test AFTER collecting all groups
    stat_levene, p_value_levene = stats.levene(*groups_for_levene)
    print(f"   Homogeneity of Variances (Levene) on Residuals: p-value = {p_value_levene:.4f}")

    #conclusion
    print("\n Conclusion on Interaction:")
    if interaction_p_value < 0.05:
        print(f"   Since the interaction p-value ({interaction_p_value:.4f}) is less than 0.05, we REJECT the null hypothesis.")
        print("   Conclusion: There is a statistically significant interaction effect between Dentist and Method.")

        # 5. Identify differences (Post-hoc Tukey HSD)
        print("\n Identifying Significant Differences (Tukey HSD on Interaction Groups):")

        # Calculate Tukey HSD on the combined interaction groups
        tukey = pairwise_tukeyhsd(endog=subset['Response'], groups=subset['Interaction_Group'], alpha=0.05)
        print(tukey)

    else:
        print(f"   Since the interaction p-value ({interaction_p_value:.4f}) is greater than 0.05, we FAIL TO REJECT the null hypothesis.")
        print("   Conclusion: There is NO statistically significant interaction effect between Dentist and Method.")
        print("   Identifying Differences: Not necessary, as the overall interaction is not significant.")
```

**Key Insights**
1. **Dentist has a significant impact on hardness:** Implant quality is operator-dependent, not just process-driven.

2. **Method significantly influences hardness**:The choice of method is critical for achieving optimal implant performance.

3. **Strong interaction between Dentist and Method**: There is no universally best method — performance depends on who uses it.

4. **Combined effect is highly important**:The system behaves like a custom-fit model, not a one-size-fits-all process.

5. **Variability exists in performance**: Lack of consistency can affect implant reliability and patient safety.
